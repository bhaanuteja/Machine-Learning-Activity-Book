```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame({

    'Variable': [
        'Climate',
        'Housing',
        'Health',
        'Crime',
        'Transportation',
        'Education',
        'Arts',
        'Recreation',
        'Economy'
    ],

    'PC1': [
        0.190,
        0.544,
        0.782,
        0.365,
        0.585,
        0.394,
        0.985,
        0.520,
        0.142
    ],

    'PC2': [
        0.017,
        0.020,
        -0.605,
        0.294,
        0.085,
        -0.273,
        0.126,
        0.402,
        0.150
    ],

    'PC3': [
        0.207,
        0.204,
        0.144,
        0.585,
        0.234,
        0.027,
        -0.111,
        0.519,
        0.239
    ]

})

data
```

```python
print("Shape:", data.shape)

print("\nMissing Values:")
print(data.isnull().sum())

data
```

```python
results = []

for pc in ['PC1', 'PC2', 'PC3']:

    index = data[pc].abs().idxmax()

    results.append({

        'Principal Component': pc,

        'Dominant Variable':
            data.loc[index, 'Variable'],

        'Loading':
            data.loc[index, pc],

        'Absolute Loading':
            abs(data.loc[index, pc])
    })


results_df = pd.DataFrame(results)

results_df
```

```python
for _, row in results_df.iterrows():

    print(
        row['Principal Component'],
        "→",
        row['Dominant Variable'],
        "| Loading:",
        row['Loading']
    )
```

```python
plot_data = data.set_index(
    'Variable'
)[['PC1', 'PC2', 'PC3']]

plot_data.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title(
    "PCA Loadings of Variables"
)

plt.xlabel(
    "Variables"
)

plt.ylabel(
    "Loading"
)

plt.xticks(
    rotation=45
)

plt.axhline(
    0,
    linewidth=1
)

plt.legend(
    title="Principal Component"
)

plt.show()
```
